```python
# Individual Model Initializations
Svc_model = SVC()
Knn_model = KNeighborsClassifier()
Gnb_model = GaussianNB()
Sgd_model = SGDClassifier()
Gbc_model = GradientBoostingClassifier()
Cat_model = CatBoostClassifier(verbose=0) # verbose=0 stops long training logs
Abc_model = AdaBoostClassifier()

# Adding the standard ones from your regression list (converted to classification)
Logr_model = LogisticRegression()
Dtc_model = DecisionTreeClassifier()
Rfc_model = RandomForestClassifier()
Xgb_model = XGBClassifier()

# Creating the models list
models = [
    Svc_model, 
    Knn_model, 
    Gnb_model, 
    Sgd_model, 
    Gbc_model, 
    Cat_model, 
    Abc_model, 
    Logr_model, 
    Dtc_model, 
    Rfc_model, 
    Xgb_model
]

# Display the list
models
```

```python
for m in models:
    m.fit(X_train_res, y_train_res)
    s = m.score(X_test, y_test)
    print(m, ":", s)
    time.sleep(2)
```

```python
# 1. Initialize an empty list to store our "Tournament" results
results_list = []

# 2. Master Loop
for model in models:
    model_name = str(model).split('(')[0] # Extracts just the name like 'SVC'
    scores = []
    
    for i in range(0, 100): 
        # A. Split the original data (X and y)
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=i)
        
        # B. Oversample ONLY the training portion (Prevents Leakage!)
        ros = RandomOverSampler(random_state=42)
        X_train_res, y_train_res = ros.fit_resample(X_train, y_train)
        
        # C. Train and Score
        model.fit(X_train_res, y_train_res)
        sc = model.score(X_test, y_test)
        scores.append(sc)
    
    # 3. Save the best results for this specific model
    max_sc = max(scores)
    best_state = scores.index(max_sc)
    avg_sc = sum(scores) / len(scores) # Added this for your Resume!
    
    results_list.append({
        'Model': model_name,
        'Max_Accuracy': max_sc * 100,
        'Avg_Accuracy': avg_sc * 100,
        'Best_Random_State': best_state
    })
    
    print(f"Finished: {model_name} | Best: {max_sc*100:.2f}%")

# 4. Convert to a Table and show the winners
summary_df = pd.DataFrame(results_list)
print("\n--- TOURNAMENT SUMMARY ---")
print(summary_df.sort_values(by='Max_Accuracy', ascending=False))
```
